```markdown
**計算規則**：
1. 服務費必須以「原始餐費」計算，不能先扣除折價券。
2. 應付總金額 = 原始餐費 + 服務費 - 折價券。
3. 每人平分費用 = 應付總金額 / 聚餐人數。
```

### 聚餐費用試算器教學與步驟說明

我們將這個計算器拆解為以下幾個步驟，讓程式邏輯清晰好懂：

1. **文字轉數字 (Type Conversion)**：
   從輸入或設定得到的餐費與折價券通常是字串（文字），必須使用 `int()`（整數）或 `float()`（浮點數，即帶有小數點的數字）將其轉換，否則系統會無法進行數學運算。例如：`"1000"` 只是文字，`int("1000")` 才會變成可以計算的數字 `1000`。

2. **計算順序 (Order of Operations)**：
   根據您的規則，計算順序為：
   * **步驟 A**：計算服務費。服務費必須以「原始餐費」計算（例如 10% 服務費 = 原始餐費 $\times$ 0.1）。
   * **步驟 B**：計算總金額。總金額 = 原始餐費 + 服務費 - 折價券。
   * **步驟 C**：計算每人平分費用。平分費用 = 總金額 / 總人數。

3. **f-string 格式化說明**：
   我們使用 `f"字串 {變數}"` 來輸出結果。如果遇到小數，我們可以使用 `{變數:.1f}`，這代表將數字格式化為**浮點數（f），並保留到小數點後第 1 位（.1）**。

### 1. 請先在心中或紙上預測以下設定的計算結果：
* **原始餐費**：`"2500"` 元
* **服務費率**：`10%` (0.1)
* **折價券**：`"200"` 元
* **聚餐人數**：`4` 人

*您的預測總金額是多少？每人平均要分攤多少？*
準備好後，請執行下方的程式碼來檢查您的預測是否正確！

````python
# ---- 步驟 1：設定原始資料（皆為基本字串與數字） ----
original_price_str = "2500"  # 原始餐費（字串）
coupon_str = "200"           # 折價券金額（字串）
service_rate = 0.1           # 服務費率 10%（數字）
people_count = 4             # 聚餐人數（數字）

# ---- 步驟 2：資料整理（將字串轉換成可計算的數字） ----
# 使用 int() 將餐費與折價券從字串轉為整數
original_price = int(original_price_str)
coupon = int(coupon_str)

# ---- 步驟 3：費用計算（步驟分開，確保邏輯清晰） ----
# 1. 計算服務費（使用原始餐費計算）
service_charge = original_price * service_rate

# 2. 計算總金額（原始餐費 + 服務費 - 折價券）
total_amount = original_price + service_charge - coupon

# 3. 計算每人平分費用
price_per_person = total_amount / people_count

# ---- 步驟 4：檢查計算結果是否合理（布林值判斷） ----
is_valid_calculation = total_amount > 0

# ---- 步驟 5：輸出結果（使用 f-string 格式化） ----
print("========= 聚餐費用試算結果 =========")
print(f"原始餐費：{original_price} 元")
print(f"服務費 ({service_rate * 100:.0f}%)：{service_charge:.0f} 元")
print(f"折價券折抵：{coupon} 元")
print(f"-----------------------------------")
print(f"計算結果是否有效（總金額大於零）：{is_valid_calculation}")
print(f"應付總金額：{total_amount:.0f} 元")
print(f"平分人數：{people_count} 人")
print(f"每人應付：{price_per_person:.1f} 元")  # :.1f 代表顯示到小數點後第一位
print("===================================")
````

```markdown
為了符合作業要求，以下我們設計了三組不同的測試資料。請依序執行下方的測試儲存格，以產生實際的執行輸出結果。
```

```markdown
### 測試案例一：一般聚餐（預設組）
* 原始餐費：2500 元
* 服務費率：10% (0.1)
* 折價券：200 元
* 人數：4 人
```

````python
# 測試案例一 程式碼
original_price_str1 = "2500"
coupon_str1 = "200"
service_rate1 = 0.1
people_count1 = 4

# 計算
original_price1 = int(original_price_str1)
coupon1 = int(coupon_str1)
service_charge1 = original_price1 * service_rate1
total_amount1 = original_price1 + service_charge1 - coupon1
price_per_person1 = total_amount1 / people_count1
is_valid1 = total_amount1 > 0

# 輸出
print("========= 測試案例一 =========")
print(f"原始餐費：{original_price1} 元")
print(f"服務費：{service_charge1:.0f} 元")
print(f"折價券折抵：{coupon1} 元")
print(f"總金額：{total_amount1:.0f} 元")
print(f"每人分攤：{price_per_person1:.1f} 元")
print(f"有效性檢查：{is_valid1}")
print("=============================")
````
